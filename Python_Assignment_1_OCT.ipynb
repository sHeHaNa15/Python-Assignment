# Python Assignment 1: Strings & Tuples

## 1. String Concatenation

```python
s1 = "Hello "
s2 = input("Shihana: ")
greeting = s1 + s2
print(greeting)

s3 = ", welcome to Python programming"
full = greeting + s3
print(full)
```

## 2. String Slicing and Indexing

```python
print("a. First character:", full[0])
print("b. Last character:", full[-1])
print("c. First 5 characters:", full[:5])
print("d. Last 11 characters:", full[-11:])
print("e. Reversed:", full[::-1])

start = full.index("Python")
print("f. Word 'Python':", full[start:start + 6])
```

## 3. String Methods

```python
strM = "Python beginner tutorial"

print("a. Uppercase:", strM.upper())
print("b. Lowercase:", strM.lower())
print("c. Capitalize:", strM.lower().capitalize())
print("d. Count of 't':", strM.count("t"))
print("e. Replaced:", strM.replace("Python", "Machine Learning"))
```

## Tuples

```python
t1 = (10, 20, 30)
t2 = (40, 50, 60)

t_combine = t1 + t2
print("a. Combined:", t_combine)
print("b. Repeated 3 times:", t_combine * 3)
print("c. 3rd element:", t_combine[2])
print("d. First three elements:", t_combine[:3])
print("e. Last three elements:", t_combine[-3:])
```
